# 04 — Azure ML Deployment
**PanganTrace AI** — Register & Deploy Models ke Azure ML

Notebook ini mengemas model terlatih dan mendeploy ke Azure ML Managed Endpoints.

```python
import sys
sys.path.insert(0, '../src')
import os
```

## 1. Setup Azure ML Connection

```python
from azure_ml_pipeline import AzureMLPipeline

pipeline = AzureMLPipeline(
    subscription_id=os.getenv('AZURE_SUBSCRIPTION_ID'),
    resource_group=os.getenv('AZURE_RESOURCE_GROUP'),
    workspace_name=os.getenv('AZURE_ML_WORKSPACE', 'pangantrace-ml'),
)

ml_client = pipeline.get_workspace()
```

## 2. Register Fraud Model

```python
fraud_result = pipeline.register_fraud_model(
    model_path='../models/fraud_model.pkl',
    model_name='pangantrace-fraud'
)
print(fraud_result)
```

## 3. Register Price Model

```python
price_result = pipeline.register_price_model(
    model_path='../models/price_model.pkl',
    model_name='pangantrace-price'
)
print(price_result)
```

## 4. Endpoint Configuration

```python
fraud_config = pipeline.create_endpoint_config('fraud')
print('Fraud Endpoint Config:')
for k, v in fraud_config.items():
    print(f'  {k}: {v}')

print()

price_config = pipeline.create_endpoint_config('price')
print('Price Endpoint Config:')
for k, v in price_config.items():
    print(f'  {k}: {v}')
```
